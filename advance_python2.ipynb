First Non-Repeating Character

```python
from collections import Counter
```

```python
def non_rep(t):
    counts=Counter(t)
    for char in t:
        if counts[char]==1:
           return char
txt="cxyaafffeetfdae"
non_rep(txt)
        
```

Longest Consecutive Sequence

```python
def long_con(num):
    num_set=set(num)
    long=0
    best=[]
    for nums in num_set:
        if nums -1 not in num_set:
            cur=nums
            seq=[]
            while cur in num_set:
                seq.append(cur)
                cur+=1
            if len(seq)>long:
                long=len(seq)
                best=seq
    print(best) 
    print(long)   
n=[100,101,104,105,103,102,106, 4, 200, 1, 3, 2]  
long_con(n)           
```
