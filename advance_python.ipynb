Find All Pairs With a Given Sum

```python
def sum_p(t):
    nums  = [2, 4, 3, 5, 7, 8, 9, 1]
    seen=set()
    pairs=set()

    for num in nums:
        c=t-num
        if c in seen:
            pair=tuple (sorted((num,c)))
            pairs.add(pair)
        seen.add(num)
        
    for a,b in pairs:
        print(a,"+",b,"=",t)
    print(pairs)  
    print(seen) 


   
sum_p(10)

```
